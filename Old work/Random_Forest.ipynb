```python
# Random Forest Regression
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv('Oil_Well.csv')
x = df.iloc[:, [5, 8]].values
y = df.iloc[:, 1].values
```

```python
# Split into training and test sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
df.head()
```

```python
# Train the model
Powei = RandomForestRegressor(n_estimators=100, max_depth=None, min_samples_split=2, min_samples_leaf=1, random_state=42, n_jobs=-1)
Powei.fit(x_train, y_train)
```

```python
 # Predict
y_pred = Powei.predict(x_test)
```

```python
# Evaluate
mae = mean_absolute_error(y_test, y_pred)
mse  = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2   = r2_score(y_test, y_pred)

print(f"Estimators   : {Powei.n_estimators}")
print(f"Max Depth    : {Powei.max_depth}")
print(f"Min Samples Split : {Powei.min_samples_split}")
print(f"Min Samples Leaf  : {Powei.min_samples_leaf}")
print(f"MAE          : {mae:.4f}")
print(f"MSE          : {mse:.4f}")
print(f"RMSE         : {rmse:.4f}")
print(f"R² Score     : {r2:.4f}")
```

```python
# Feature importances
feat_names   = df.columns[[5, 8]].tolist()
importances  = Powei.feature_importances_
std          = np.std([tree.feature_importances_ for tree in Powei.estimators_], axis=0)
indices      = np.argsort(importances)[::-1]
plt.figure(figsize=(10, 5))
plt.bar(range(len(importances)), importances[indices], yerr=std[indices], color='grey', alpha=0.8, capsize=4)
plt.xticks(range(len(importances)), [feat_names[i] for i in indices], rotation=45, ha='right')
plt.title("Random Forest - Feature Importances (with std)")
plt.tight_layout()
plt.savefig("rf_feature_importance.png", dpi=150)
plt.show()
```

```python
# Actual vs Predicted
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred, color='black', alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Random Forest Regressor – Actual vs Predicted")
plt.tight_layout()
plt.savefig("rf_actual_vs_predicted.png", dpi=150)
plt.show()
```
