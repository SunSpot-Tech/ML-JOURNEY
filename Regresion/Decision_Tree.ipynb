```python
#Decision Tree Regression
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, export_text, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
#Load dataset
df = pd.read_csv('Oil_Well.csv')
x = df.iloc[:, [5, 8]].values
y = df.iloc[:, 1].values
```

```python
#Split dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#Train Decision Tree model
Powei = DecisionTreeRegressor(max_depth=5, min_samples_split=5, random_state=42)
Powei.fit(x_train, y_train)
```

```python
#Predicting the Test set results
y_pred = Powei.predict(x_test)
```

```python
#Evaluate the model
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rsme = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)


print(f"Max Depth: {Powei.max_depth}")
print(f"Tree Depth: {Powei.get_depth()}")
print(f"Leaves: {Powei.get_n_leaves()}")
print(f"mae: {mae:.4f}")
print(f"mse: {mse:.4f}")
print(f"rsme: {rsme:.4f}")
print(f"r2: {r2:.4f}")
```

```python
#Feature importances
feat_names = df.columns[[5, 8]].tolist()
importances = Powei.feature_importances_
indices = np.argsort(importances)[::-1]
plt.figure(figsize=(10, 5))
plt.bar(range(len(importances)), importances[indices], color='green', alpha=0.7, align='center')
plt.xticks(range(len(importances)), [feat_names[i] for i in indices], rotation=45, ha = 'right')
plt.title('Decision Tree - Feature Importances')
plt.tight_layout()
plt.savefig('Decision_Tree_Feature_Importances.png', dpi=150)
plt.show()
```
