Seperating the features (x) and target (y) variables

```python
x = df.iloc[:, [5]].values
y = df.iloc[:, 1].values
```

```python
print(x)
```

```python
print(y)
```

No encoding needed, since my data is not having categorical data or strings.
No filliing of data needed, since there are no missing data.
Splitting my data. 

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
print(x_train)
```

```python
print(x_test)
```

```python
print(y_train)
```

```python
print(y_test)
```

Scaling my data

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
print(x_train)
```

```python
print(x_test)
```

Building my model

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

Testing my model

```python
y_pred = model.predict(x_test)
print(y_pred)
```

Mean Absolute Error (MAE)

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test, y_pred)
print(mae)
```

Mean Squared Error (MSE)

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
print(mse)
```

Checking r2 Score

```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test, y_pred)
print(r2)
```

From the r2 score, the model should be able to interpret up to 90% of my data. With the remaining 10% due to factors we haven't accounted for like Working hours, Dynamic level etc.
